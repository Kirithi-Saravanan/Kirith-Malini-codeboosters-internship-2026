### Dataset Overview

```python
print(f"Total number of columns: {df.shape[1]}")
print(f"Total number of students: {df.shape[0]}")
print(f"Number of unique departments: {df['department'].nunique()}")
```

### Student Count by Department

```python
print("Count of students in each department:")
print(df['department'].value_counts())
```

### Score Analysis

```python
print(f"Highest Math Score: {df['math_score'].max()}")
print(f"Lowest Math Score: {df['math_score'].min()}")
print(f"Average Math Score: {df['math_score'].mean():.2f}")
print(f"Average Programming Score: {df['programming_score'].mean():.2f}")
print(f"Average Attendance Percentage: {df['attendance_percentage'].mean():.2f}%")
```

### Average Total Score by Gender

```python
print("Average total score grouped by gender:")
print(df.groupby('gender')['total_score'].mean())
```

### Top 5 Scorers (by Total Score)

```python
top_5_scorers = df.sort_values(by='total_score', ascending=False).head(5)
print("Top 5 performers based on total scores:")
display(top_5_scorers[['name', 'department', 'total_score']])
```

### Students with Attendance Below 75%

```python
low_attendance_students = df[df['attendance_percentage'] < 75]

if not low_attendance_students.empty:
    print("Students with attendance below 75%:")
    display(low_attendance_students[['name', 'department', 'attendance_percentage']])
else:
    print("All students have an attendance percentage of 75% or higher.")
```

```python
electronics_students = df[df['department'] == 'Electronics']
print(f"Number of students from Electronics department: {electronics_students.shape[0]}")
```

```python
female_students = df[df['gender'] == 'Female']
average_female_attendance = female_students['attendance_percentage'].mean()
print(f"Average attendance of female students: {average_female_attendance:.2f}%")
```

```python
lowest_programming_student = df.loc[df['programming_score'].idxmin()]
print(f"Student with the lowest programming score: {lowest_programming_student['name']}")
```

```python
high_math_scores = df[df['math_score'] > 80]
print(f"Number of students who scored above 80 in math: {high_math_scores.shape[0]}")
```

```python
high_attendance_students = df[df['attendance_percentage'] > 90]
average_total_score_high_attendance = high_attendance_students['total_score'].mean()
print(f"Average total score for students with attendance above 90%: {average_total_score_high_attendance:.2f}")
```
