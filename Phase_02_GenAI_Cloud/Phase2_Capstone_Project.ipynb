​Tech Stack: Pandas + Groq API

```python
import pandas as pd

df = pd.read_csv("student_performance.csv")

print(df.head())
print(df.info())
```

```python
print("Average GPA:", df["gpa"].mean())
```

```python
print(
    df.groupby("branch")["gpa"]
    .mean()
    .sort_values(ascending=False)
)
```

```python
pass_rate = (
    df["passed"].value_counts(normalize=True)
    *100
)

print(pass_rate)
```

```python
import sqlite3

conn = sqlite3.connect("student.db")

df.to_sql(
    "students",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT branch,
AVG(gpa) as avg_gpa
FROM students
GROUP BY branch
ORDER BY avg_gpa DESC
"""

result = pd.read_sql(query, conn)

print(result)
```
